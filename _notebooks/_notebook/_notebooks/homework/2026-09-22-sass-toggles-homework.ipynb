# SASS Toggles — Homework

Completed homework based directly on the uploaded lesson requirements.

```python
from IPython.display import HTML, display

html = '''
<div>
  <h2>Interactive Toggle Dashboard</h2>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="notifications" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Notifications</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="dark-mode" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Dark Mode</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="privacy" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Privacy Mode</span>
  </label>

  <p id="status">0 toggles are currently turned on.</p>
  <p id="feature">Notifications are off.</p>

  <script>
    const toggles = [
      document.getElementById("notifications"),
      document.getElementById("dark-mode"),
      document.getElementById("privacy")
    ];

    const status = document.getElementById("status");
    const feature = document.getElementById("feature");

    function updateDashboard() {
      const onCount = toggles.filter(toggle => toggle.checked).length;
      status.textContent = `${onCount} toggle${onCount === 1 ? "" : "s"} are currently turned on.`;

      const notifications = document.getElementById("notifications");
      feature.textContent = notifications.checked
        ? "Notifications are on."
        : "Notifications are off.";
    }

    toggles.forEach(toggle => {
      toggle.addEventListener("change", updateDashboard);
    });

    // I thought .checked was useful because JavaScript can directly tell
    // whether each checkbox is currently turned on or off.
    // The counter updates every time a toggle changes, so the page reflects
    // the current state instead of using a fixed number.
    updateDashboard();
  </script>
</div>
'''

display(HTML(html))
```

## Requirements completed

- 3 toggles using `ocs__toggle`
- Each toggle has its own label and purpose
- JavaScript checks whether each toggle is on or off
- The page updates based on toggle status
- A live count shows how many toggles are on
- The count updates when toggles are flipped
- Comments explain understanding of `.checked` and the live counter
- The Notifications toggle actually controls a page feature
