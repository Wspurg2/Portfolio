```python
# CODE_RUNNER:

import json
import statistics
from datetime import date

games = [
    {"date": "2026-09-14", "opponent": "Lincoln", "points": 24, "minutes": 28},
    {"date": "2026-09-16", "opponent": "Madison", "points": 18, "minutes": 22},
    {"date": "2026-09-18", "opponent": "Roosevelt", "points": 31, "minutes": 30},
    {"date": "2026-09-21", "opponent": "Lincoln", "points": 27, "minutes": 26}
]

def summarize_season(season):
    """Summarize a list of games.

    Parameters:
        season: list of game dictionaries

    Returns:
        dictionary with total points, average points, best game,
        and points by opponent
    """
    total_points = sum(game["points"] for game in season)
    average_points = statistics.mean(game["points"] for game in season)
    best_game = max(season, key=lambda game: game["points"])

    points_by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["points"]

    return {
        "total_points": total_points,
        "average_points": round(average_points, 2),
        "best_game": best_game,
        "points_by_opponent": points_by_opponent
    }

report = summarize_season(games)
print(json.dumps(report, indent=2))

next_game = date(2026, 10, 15)
days_until_game = (next_game - date.today()).days
print("Days until next game:", days_until_game)
```
