# OCS SASS Buttons Grammar — Homework Hacks

Completed **Homework Hack #1** and **Homework Hack #2** using the required OCS button and layout classes.

## Homework Hack #1 — Adaptive Resource Toolbar

```python
from IPython.display import HTML, display

homework_1 = '''
<!-- UI_RUNNER: Homework Hack #1 - Adaptive Resource Toolbar -->
<div class="ocs__links">
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-red fill pill">Urgent Support</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-green fill pill">Self-Care</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn accent outline pill">Community</a>
</div>
'''

display(HTML(homework_1))
```

This uses `ocs__links` as the action-group container and removes the starter's inline styles and hardcoded colors. Each resource receives the exact semantic OCS classes required by the assignment.

## Homework Hack #2 — Primary vs. Secondary Action Priority

```python
homework_2 = '''
<!-- UI_RUNNER: Homework Hack #2 - Visual Action Hierarchy -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn outline">View Weekly Trends</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-red outline">Crisis Hotline</a>
</div>
'''

display(HTML(homework_2))
```

The primary action uses `accent fill`, the secondary action uses `outline`, and the alert action uses `alert-red outline`, all inside `ocs__links`.
