# 3.4 Strings Homework

**Topic:** Strings — indexing, slicing, string methods, concatenation, and formatting

**Learning Objective:** I can use indexing and slicing to get parts of a string, use string methods to clean and search text, and combine strings with concatenation and f-strings.

## Popcorn Hack 1: Extract the Hidden Message

Use indexing, slicing, negative indexing, and `len()` to decode the transmission.

```python
# CODE_RUNNER: Popcorn Hack 1 - Extract the hidden message using indexing and slicing

transmission = "xxSIGNALxxxCONFIRMEDxx77xxDELTA"

# TODO 1: Pull out "SIGNAL" (starts at index 2 and is 6 characters long)
word_1 = transmission[2:8]
print("Word 1:", word_1)

# TODO 2: Pull out "CONFIRMED"
word_2 = transmission[11:20]
print("Word 2:", word_2)

# TODO 3: Use negative indexing/slicing to pull out the last 5 characters
word_3 = transmission[-5:]
print("Word 3:", word_3)

# TODO 4: Print the number of characters in the full transmission
print("Total length:", len(transmission))

# TODO 5: Print every other character
print("Every other character:", transmission[::2])
```

## Popcorn Hack 2: Scrub and Search a Noisy Transmission

Clean the message by removing extra whitespace and punctuation, convert it to lowercase, search for a keyword, and count its words.

```python
# CODE_RUNNER: Popcorn Hack 2 - Clean and search a noisy transmission using string methods

noisy = "   ...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!   "

# TODO 1: Remove leading and trailing whitespace
trimmed = noisy.strip()
print("Trimmed:", repr(trimmed))

# TODO 2: Remove every ">", "-", and "!" character
cleaned = trimmed.replace(">", "").replace("-", "").replace("!", "")
print("Cleaned:", cleaned)

# TODO 3: Make a lowercase copy
lowered = cleaned.lower()
print("Lowered:", lowered)

# TODO 4: Check whether "arrived" appears in the message
has_arrived = "arrived" in lowered
print("Contains 'arrived':", has_arrived)

# TODO 5: Split into words and print the number of words
words = cleaned.split()
print("Words:", words)
print("Word count:", len(words))
```

## Popcorn Hack 3: Build a Decoded Report

Use concatenation, `str()`, f-strings, and `.join()` to combine the intercepted information.

```python
# CODE_RUNNER: Popcorn Hack 3 - Build a decoded report using concatenation and formatting

agent_name = "Falcon"
target_word = "MIDNIGHT"
confidence = 87
fragments = ["ALPHA", "SEVEN", "DELTA"]

# TODO 1: Build the headline with + concatenation
headline = "Agent " + agent_name + " intercepted: " + target_word
print(headline)

# TODO 2: Use str() and + concatenation to include the confidence number
confidence_line = "Confidence: " + str(confidence) + "%"
print(confidence_line)

# TODO 3: Build the same line using an f-string
confidence_line_fstring = f"Confidence: {confidence}%"
print(confidence_line_fstring)

# TODO 4: Join the fragments with dashes
joined_fragments = "-".join(fragments)
print("Joined fragments:", joined_fragments)
```

## Homework: Crack the Final Transmission

Decode the CSV-style transmission. The program cleans and splits the data, extracts each field, checks the year, searches for the word “dawn,” and rebuilds the information as a Markdown table row and a JSON-style line.

```python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# TODO 1: Strip outer whitespace and split the CSV row into fields
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# TODO 2: Use indexing to get the date, agent, message code, and confidence
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

# TODO 3: Use slicing to check whether the date starts with "2026"
year_check = date[:4] == "2026"
print("Year check:", year_check)

# TODO 4: Replace the dashes in the message code with spaces
message = message_code.replace("-", " ")

# TODO 5: Search for "dawn" without worrying about capitalization
found_dawn = "dawn" in message.lower()
print("Decoded message:", message)
print("Found 'dawn':", found_dawn)

# TODO 6: Build a Markdown table row with an f-string
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

# TODO 7: Build a JSON-style line. Convert the Boolean to lowercase for JSON.
json_line = f'{{"date": "{date}", "agent": "{agent}", "message": "{message}", "confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}}}'
print("JSON line:", json_line)
```

## Reflection

Strings are immutable, which means their contents cannot be changed in place. Indexing and slicing let me access parts of a string, while methods such as `.strip()`, `.replace()`, `.lower()`, and `.split()` return cleaned or transformed values. Concatenation, f-strings, and `.join()` help combine those values into a readable report.
