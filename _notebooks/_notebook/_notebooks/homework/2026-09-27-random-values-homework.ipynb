# 3.15 Random Values — Homework Hack
**SFI Backend QA Simulator**

Completed solution based on the supplied starter data and requirements.

```python
import random

parts = [
    {
        "product_name": "Replacement Flywheels",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

actions = [
    "GET/search",
    "POST/create",
    "PUT/update",
    "DELETE/remove"
]

existing_spec_numbers = ["1.1", "2.1"]
test_count = 5

def run_qa_test(part, action):
    spec = part["spec_number"]

    if action == "GET/search":
        return f"Search result: found {part['product_name']} (spec {spec})."

    if action == "POST/create":
        if spec in existing_spec_numbers:
            return f"Create rejected: spec {spec} already exists."
        return f"Create accepted: {part['product_name']} (spec {spec})."

    if action == "PUT/update":
        if spec in existing_spec_numbers:
            return f"Update accepted: existing spec {spec} can be updated."
        return f"Update rejected: spec {spec} does not exist."

    if action == "DELETE/remove":
        if spec in existing_spec_numbers:
            return f"Remove accepted: existing spec {spec} selected."
        return f"Remove rejected: spec {spec} does not exist."

    return "Unknown action."

for run_number in range(1, test_count + 1):
    selected_part = random.choice(parts)
    selected_action = random.choice(actions)
    result = run_qa_test(selected_part, selected_action)

    print(f"Run {run_number}")
    print("  Record:", selected_part)
    print("  Action:", selected_action)
    print("  Result:", result)
    print()
```

### Possible random results

`random.choice(parts)` can select any of the three complete records, and `random.choice(actions)` can select any of the four actions. Therefore there are **3 × 4 = 12 possible record/action combinations**. A combination can repeat on consecutive runs.

The create path checks `existing_spec_numbers` so duplicate-spec behavior is handled explicitly.
