# 3.07 Nested Conditionals — Homework Hack
**High School Daily Routine Dashboard**

Completed solution based on the supplied starter data and requirements.

```python
student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

if student["logged_in"]:
    if student["homework_finished"]:
        if student["practice_today"]:
            if student["lunch_packed"]:
                print("Homework finished — lunch is packed and practice is today.")
            else:
                print("Homework finished — pack or buy lunch before practice.")
        else:
            if student["lunch_packed"]:
                print("Homework finished — lunch is packed and there is no practice today.")
            else:
                print("Homework finished — arrange lunch before the rest of the day.")
    else:
        if student["practice_today"]:
            if student["lunch_packed"]:
                print("Finish homework before practice; lunch is already packed.")
            else:
                print("Finish homework before practice and arrange lunch.")
        else:
            if student["lunch_packed"]:
                print("Finish homework before free time; lunch is already packed.")
            else:
                print("Finish homework and arrange lunch before free time.")
else:
    print("Log in first.")
```

### Testing every meaningful outcome

The tests below change the four starter values so the dashboard reaches different nested branches.

```python
test_cases = [
    {"logged_in": True, "homework_finished": True, "lunch_packed": True, "practice_today": True},
    {"logged_in": True, "homework_finished": False, "lunch_packed": False, "practice_today": True},
    {"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": False},
    {"logged_in": False, "homework_finished": False, "lunch_packed": True, "practice_today": True},
]

def dashboard(student):
    if student["logged_in"]:
        if student["homework_finished"]:
            if student["practice_today"]:
                if student["lunch_packed"]:
                    return "Homework finished — lunch is packed and practice is today."
                return "Homework finished — pack or buy lunch before practice."
            if student["lunch_packed"]:
                return "Homework finished — lunch is packed and there is no practice today."
            return "Homework finished — arrange lunch before the rest of the day."
        else:
            if student["practice_today"]:
                if student["lunch_packed"]:
                    return "Finish homework before practice; lunch is already packed."
                return "Finish homework before practice and arrange lunch."
            if student["lunch_packed"]:
                return "Finish homework before free time; lunch is already packed."
            return "Finish homework and arrange lunch before free time."
    return "Log in first."

for i, case in enumerate(test_cases, 1):
    print(f"Test {i}: {dashboard(case)}")
```

The inner checks depend on the outer decisions because Python only reaches an indented inner `if` after the outer condition selects that branch. The account check comes first, so schedule recommendations are not shown when the student is not logged in.
